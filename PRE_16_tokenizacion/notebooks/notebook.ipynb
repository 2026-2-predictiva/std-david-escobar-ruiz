```python
import os

import nltk

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)

os.makedirs("../submission", exist_ok=True)
```

```python
with open("../data/file1.txt", encoding="utf-8") as file:
    text = file.read()
print(text)
```

```python
# Lines are hard-wrapped, so they are joined before splitting into sentences
def tokenize(text):
    text = " ".join(text.split())
    sentences = nltk.sent_tokenize(text)
    return [nltk.word_tokenize(sentence) for sentence in sentences]

tokenize(text)[:2]
```

```python
# One sentence per line, tokens separated by a single space
for name in ["file1.txt", "file2.txt", "file3.txt"]:
    with open(f"../data/{name}", encoding="utf-8") as file:
        sentences = tokenize(file.read())
    with open(f"../submission/{name}", "w", encoding="utf-8") as file:
        for tokens in sentences:
            file.write(" ".join(tokens) + "\n")
    print(name, len(sentences), "sentences")
```
